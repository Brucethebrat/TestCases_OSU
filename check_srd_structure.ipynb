```python
import json

with open("srd.json", "r", encoding="utf-8") as f:
    full_data = json.load(f)
    print(full_data.keys())

    routing_data = full_data["StaticRoutingData"]
    print("Routing Data keys: ", routing_data.keys())

    RD_airports = routing_data["Airports"]
    print(f"\nTotal number of airports in RoutingData: {len(RD_airports)}")
    print("RoutingData['Airports'][0] keys:", RD_airports[0].keys())

    RD_RoutingCache = routing_data["RoutingCache"]
    print("\nRoutingData['RoutingCache'] keys:", RD_RoutingCache.keys())

    RD_RoutingCache_airports = RD_RoutingCache["Airports"]
    print("\nRoutingData['RoutingCache']['Airports']:", RD_RoutingCache_airports)
    print(f"Total number of airports in RoutingCache: {len(RD_RoutingCache_airports)}")

    # RD_RoutingCache_Routes = RD_RoutingCache["Routes"]
    # print("\nRoutingData['RoutingCache']['Routes'][:2]:", RD_RoutingCache_Routes[:2])
```
